```python
import pandas as pd
# 1. Read dataset

df = pd.read_csv("employee.csv")
df = df[['Age','DailyRate','Education','MonthlyIncome','Gender']]
print("Selected Dataset:")
print(df)
```

```python
# 2.Display top 5 and bottom 5 records

print("\nTop 5 Records:")
print(df.head())
print("\nBottom 5 Records:")
print(df.tail())
```

```python
# 3. Select column with specific datatypes(Hint:df.select_dtypes)
print("\nInteger Columns:")
print(df.select_dtypes(include=['int']))

print("\nobject Columns:")
print(df.select_dtypes(include=['object']))
```

```python
# 4. Find maximum and minimum Monthly Income

print("\nMaximum Monthly Income:")
print(df['MonthlyIncome'].max())

print("\nMinimum Monthly Income:")
print(df['MonthlyIncome'].min())
```

```python
# 5. Measure of Central Tendency(Using Age column)

print("\nMean:")
print(df['Age'].mean())

print("\nMedian:")
print(df['Age'].median())

print("\nMode:")
print(df['Age'].mode()[0])
```

```python
# 6. Measure of Spread

print("\nVariance:")
print(df['Age'].var())

print("\nStandard Deviation:")
print(df['Age'].std())

print("\nRange:")
print(df['Age'].max() - df['Age'].min())

print("\nQuartiles:")
print(df['Age'].quantile([0.25, 0.50, 0.75]))

Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)

print("\nIQR:")
print(Q3 - Q1)
```

```python
# 7. Correlation Coefficient(Correlation between Age and MonthlyIncome)

correlation = df['Age'].corr(df['MonthlyIncome'])
print("\nCorrelation Coefficient between Age and MonthlyIncome:")
print(correlation)
```
