## Demonstrate exception handling using try,except,else,finally blocks and raise statement with custom exceptions.

```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num
    print("Result:", result)
except ZeroDivisionError:
    print("Error: Cannot divide by zero.")
```

```python
#else 

try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ZeroDivisionError:
    print("Division by zero is not allowed.")
    
else:
    print("Division successful.")
    print("Result:", result)
```

```python
#finally

try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ZeroDivisionError:
    print("Division by zero is not allowed.")
finally:
    print("Program finished.")    
```

```python
#raise 

age = int(input("Enter age: "))

if age < 18:
    raise ValueError("Age must be 18 or above.")

print("Eligible to vote.")
```

```python
#custom exception : make your own exception class

class InvalidMarkError(Exception):
    pass

mark = int(input("Enter mark: "))

if mark < 0 or mark > 100:
    raise InvalidMarkError("mark must be between 0 and 100.")

else:
    print("Mark
          :", mark)
```

```python
class InvalidAgeError(Exception):
    pass

age = int(input("Enter age: "))

if age < 18:
    raise InvalidAgeError("You must be at least 18 years old.")

print("Access granted.")
```
