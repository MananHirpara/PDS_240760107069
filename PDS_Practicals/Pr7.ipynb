## Connect python with MySQL database and perfrom create,read,update and delete(CRUD)operation.

```python
!pip install mysql-connector-python
import mysql.connector

db = mysql.connector.connect(
    host="localhost",
    user="root",
)
cursor = db.cursor()

cursor.execute("CREATE DATABASE IF NOT EXISTS PDS")
cursor.execute("USE PDS")

cursor.execute("""
CREATE TABLE IF NOT EXISTS users (
    id INT AUTO_INCREMENT PRIMARY KEY,
    name VARCHAR(255),
    email VARCHAR(255)
)
""")

sql_create = "INSERT INTO users (name, email) VALUES (%s, %s)"
val_create = ("John Doe", "john@example.com")
cursor.execute(sql_create, val_create)
db.commit()
print(f"Created: {cursor.rowcount} row inserted. ID: {cursor.lastrowid}")

sql_read = "SELECT id, name, email FROM users"
cursor.execute(sql_read)
results = cursor.fetchall()
print("Reading current rows:")
for row in results:
    print(row)

sql_update = "UPDATE users SET email = %s WHERE name = %s"
val_update = ("john.new@example.com", "John Doe")
cursor.execute(sql_update, val_update)
db.commit()
print(f"Updated: {cursor.rowcount} row(s) affected.")

sql_delete = "DELETE FROM users WHERE name = %s"
val_delete = ("John Doe",)
cursor.execute(sql_delete, val_delete)
db.commit()
print(f"Deleted: {cursor.rowcount} row(s) affected.")

cursor.close()
db.close()
import mysql.connector
```

```python
# Database Connection
db = mysql.connector.connect(
    host="localhost",
    user="root",
    password="",
    database="mann069"
)
cursor = db.cursor()
print('Connected to MySQL')
```

```python
# Insert
def insert_data():
    name = input("Enter Name: ")
    age = int(input("Enter Age: "))
    course = input("Enter Course: ")

    sql = "INSERT INTO P1 (name, age, course) VALUES (%s, %s, %s)"
    val = (name, age, course)

    cursor.execute(sql, val)
    db.commit()

    print("Record Inserted Successfully!")
```

```python
# Display
def display_data():
    cursor.execute("SELECT * FROM P1")
    records = cursor.fetchall()

    print("\n----- Student Records -----")
    print("ID\tName\tAge\tCourse")
    print("-" * 35)

    for row in records:
        print(f"{row[0]}\t{row[1]}\t{row[2]}\t{row[3]}")
```

```python
# Update
def update_data():
    sid = int(input("Enter Student ID: "))

    print("\nWhich field do you want to update?")
    print("1. Name")
    print("2. Age")
    print("3. Course")

    ch = input("Enter Choice: ")

    if ch == "1":
        new_name = input("Enter New Name: ")
        sql = "UPDATE P1 SET name=%s WHERE id=%s"
        val = (new_name, sid)

    elif ch == "2":
        new_age = int(input("Enter New Age: "))
        sql = "UPDATE P1 SET age=%s WHERE id=%s"
        val = (new_age, sid)

    elif ch == "3":
        new_course = input("Enter New Course: ")
        sql = "UPDATE P1 SET course=%s WHERE id=%s"
        val = (new_course, sid)

    else:
        print("Invalid Choice!")
        return

    cursor.execute(sql, val)
    db.commit()

    if cursor.rowcount > 0:
        print("Record Updated Successfully!")
    else:
        print("No Record Found!")
```

```python
# Delete
def delete_data():
    sid = int(input("Enter Student ID to Delete: "))

    sql = "DELETE FROM P1 WHERE id=%s"
    val = (sid,)

    cursor.execute(sql, val)
    db.commit()

    if cursor.rowcount > 0:
        print("Record Deleted Successfully!")
    else:
        print("No Record Found!")
```

```python
# Main Menu
while True:
    print("\n========== CRUD MENU ==========")
    print("1. Insert Record")
    print("2. Display Records")
    print("3. Update Record")
    print("4. Delete Record")
    print("5. Exit")

    choice = input("Enter Your Choice: ")

    if choice == "1":
        insert_data()

    elif choice == "2":
        display_data()

    elif choice == "3":
        update_data()

    elif choice == "4":
        delete_data()

    elif choice == "5":
        print("Thank You!")
        break

    else:
        print("Invalid Choice!")

cursor.close()
db.close()
```
