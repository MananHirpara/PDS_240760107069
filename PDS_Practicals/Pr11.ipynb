**Aim:** Use Pandas to create Series and DataFrames, perform data selection, filtering, grouping and aggregation operations.

### 1. Creating Series and DataFrame

```python
import pandas as pd

# Series
prices = pd.Series([45000, 800, 1500, 22000, 1800], index=["Laptop", "Mouse", "Keyboard", "Monitor", "Headset"])
print("Price Series:\n", prices)

# DataFrame (Inventory data)
product_data = {
    "ProductID": ["P101", "P102", "P103", "P104", "P105", "P106"],
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Printer", "Webcam"],
    "Category": ["Electronics", "Accessories", "Accessories", "Electronics", "Electronics", "Accessories"],
    "Price": [55000, 650, 1200, 15000, 12000, 2500],
    "Stock": [15, 50, 40, 20, 10, 30]
}

df = pd.DataFrame(product_data)
print("\nProduct Inventory DataFrame:\n", df)
```

### 2. Data Selection

```python
# Selecting columns
print("Product List:\n", df["Product"])
print("\nProduct & Price:\n", df[["Product", "Price"]])

# iloc selection
print("\nFirst 4 products (iloc):\n", df.iloc[0:4])
print("\nProduct at index 3 (iloc):\n", df.iloc[3, [1, 3, 4]])

# loc selection
print("\nProducts 1 to 4 with Product & Category (loc):\n", df.loc[1:4, ["Product", "Category"]])
```

### 3. Data Filtering

```python
# Products with Price > 5000
print("High value products (Price > 5000):\n", df[df["Price"] > 5000])

# Accessories with Stock >= 35
print("\nAccessories with Stock >= 35:\n", df[(df["Category"] == "Accessories") & (df["Stock"] >= 35)])
```

### 4. Grouping and Aggregation

```python
# Mean price per category
print("Average Price by Category:\n", df.groupby("Category")["Price"].mean())

# Multi-metric aggregation
print("\nCategory-wise Stats:\n", df.groupby("Category").agg({
    "Price": ["mean", "max"],
    "Stock": ["sum", "mean"],
    "ProductID": "count"
}))
```
