**Aim:** Use SciPy's stats module to compute statistical measures and perform statistical tests on datasets.

### 1. Statistical Measures using SciPy

```python
from scipy import stats
import numpy as np

# Create Dataset (Monthly Salaries in Thousands)
salaries = [28, 32, 35, 35, 42, 50, 55, 60, 68]
print("Dataset:", salaries)

# Measures of Central Tendency (Mean, Median, Mode)
print("Mean               :", stats.tmean(salaries))
print("Median             :", np.median(salaries))
result = stats.mode(salaries)
print(f"Mode               : {result.mode} (count: {result.count})")

# Measures of Dispersion & Shape (Std Dev, Variance, Skewness, Kurtosis)
print("Standard Deviation :", round(stats.tstd(salaries), 4))
print("Variance           :", round(stats.tvar(salaries), 4))
print("Skewness           :", round(stats.skew(salaries), 4))
print("Kurtosis           :", round(stats.kurtosis(salaries), 4))

# Overall SciPy Describe Summary
print("\nSciPy Describe Summary:\n", stats.describe(salaries))
```

### 2. Probability Distributions (Normal Distribution)

```python
# Standard Normal Distribution: Probability Density Function (PDF) & Cumulative Distribution Function (CDF)
x = 1.0
print("Standard Normal PDF at x=1.0 :", round(stats.norm.pdf(x, loc=0, scale=1), 4))
print("Standard Normal CDF at Z=1.96:", round(stats.norm.cdf(1.96, loc=0, scale=1), 4))
```

### 3. Hypothesis Testing: One-Sample t-Test

```python
# Testing if average salary equals industry baseline (mu=40)
t_stat, p_val = stats.ttest_1samp(salaries, popmean=40)

print("Sample:", salaries)
print("t-statistic:", round(t_stat, 4))
print("p-value    :", round(p_val, 4))

if p_val < 0.05:
    print("Conclusion : Reject Null Hypothesis (Salary differs significantly from 40k).")
else:
    print("Conclusion : Fail to Reject Null Hypothesis (Salary is consistent with 40k baseline).")
```

### 4. Correlation Test (Pearson Correlation)

```python
# Pearson Correlation Test (Experience in Years vs Compensation)
experience = [1, 2, 4, 6, 8]
compensation = [30, 35, 50, 65, 85]

result_corr = stats.pearsonr(experience, compensation)
print("Correlation : ", round(result_corr.statistic, 4))
print("p-value     : ", round(result_corr.pvalue, 6))
```
