## Implement programs using if,if-else,elif,while,loop,for loop,nested loops, and loop control statements(break,continue,pass).

```python
age = int(input("Enter your age : "))
if age > 18:
    print("You are eligible.")
else:
    print("You are not eligible.")
```

```python
result = int(input("Enter your total marks : "))
if result > 90:
    print("Your grade is A.")
elif result > 70:
    print("Your grade is B.")
elif result > 60:
    print("Your grade is C.")
elif result > 50:
    print("Your grade is D.")
else:
    print("Guess what!!")
    print("U failed")
```

```python
password = input("Enter password : ")
if len(password) < 8 :
    print("Your password length is less then 8")
else :
    print("Password length is greater then or equal to 8")
```

```python
temp = int(input("Enter the temp : "))
if temp > 35:
    print("Warning!! should be overheating.")
```

```python
num = int(input("Enter number to check wheather its even or odd : "))
if num%2 == 0:
    print("Entered number is even.")
else :
    print("Entered number is odd.")
```

```python
i = 1
while i <= 10:
    print(i)
    i += 1
```

```python
i = 15
while i > 0:
    print(i)
    i -= 1
```

```python
for i in range(1, 11):
    print(i)
```

```python
i = 0
while i < 100:
    i += 1
    if i == 5:
        continue
    print(i)
    if i == 10:
        break
```
