# Python Programs – Variables, Data Types and Operators


## 1. Variables and Data Types


```python
a = 69
b = 6.9
c = "Hello World"
d = True
print(a, type(a))
print(b, type(b))
print(c, type(c))
print(d, type(d))
```

## 2. Arithmetic Operators


```python
x = 20
y = 10
print("Addition:", x + y)
print("Subtraction:", x - y)
print("Multiplication:", x * y)
print("Division:", x / y)
print("Modulus:", x % y)
print("Exponent:", x ** y)
print("Floor division:", x // y)
```

## 3. Assignment Operators


```python
a = 10
print("Initial a:", a)
a += 5
print("After += 5:", a)
a -= 3
print("After -= 3:", a)
a *= 2
print("After *= 2:", a)
a /= 4
print("After /= 4:", a)
a %= 4
print("After %= 4:", a)
```

## 4. Comparison Operators


```python
a = 10
b = 5
print("Is a equal to b:", a == b)
print("Is a not equal to b:", a != b)
print("Is a greater than b:", a > b)
print("Is a less than b:", a < b)
print("Is a greater than or equal to b:", a >= b)
print("Is a less than or equal to b:", a <= b)
```

## 5. Logical Operators


```python
p = True
q = False
print("P and q:", p and q)
print("P or q:", p or q)
print("not p:", not p)
```

## 6. Membership Operators


```python
nums = [1, 2, 3, 4, 5]
M = "Python"
print("3 in nums =", 3 in nums)
print("13 in nums =", 13 in nums)
print("'y' in M =", 'y' in M)
print("'m' in M =", 'm' in M)
```

## 7. Identity Operators


```python
x = [1, 2, 3]
y = [1, 2, 3]
z = x
print("x is z:", x is z)
print("x is y:", x is y)
print("x is not y:", x is not y)
```

## 8. Bitwise Operators


```python
a = 2
b = 3
print("a & b:", a & b)
print("a | b:", a | b)
print("a ^ b:", a ^ b)
print("~a:", ~a)
print("a << 1:", a << 1)
print("a >> 1:", a >> 1)
```
