**Aim:** Apply data preprocessing techniques: slicing, filtering, sorting, concatenating, aggregating and normalizing data using Pandas.

### 1. Load Dataset

```python
import pandas as pd

df = pd.read_csv('agaricus-lepiota.csv')
print('Dataset Shape:', df.shape)
df.head()
```

### 2. Slicing & Filtering

```python
# Slicing with .iloc and .loc
print('Positional Slicing (.iloc):\n', df.iloc[0:3, 0:4])
print('\nLabel Slicing (.loc):\n', df.loc[0:2, ['class', 'cap-color', 'odor']])

# Filtering
filtered = df[(df['class'] == 'e') & (df['odor'] == 'n')]
print(f'\nFiltered Count (Edible & Odor = none): {len(filtered)}')
```

### 3. Sorting & Concatenating

```python
# Sorting
print('Sorted by Odor (Descending):\n',
      df[['class', 'odor', 'habitat']].sort_values(by='odor', ascending=False).head(3))

# Concatenating (Vertical and Horizontal)
v_concat = pd.concat([df.iloc[:2, :3], df.iloc[2:4, :3]], axis=0)
print('\nVertical Concatenation:\n', v_concat)

h_concat = pd.concat([df[['class']].head(3), df[['odor']].head(3)], axis=1)
print('\nHorizontal Concatenation:\n', h_concat)
```

### 4. Aggregating & Normalizing

```python
# Create a numeric flag for aggregation
df['edible_flag'] = (df['class'] == 'e').astype(int)

# Aggregating with groupby
print('Edible Rate by Odor:\n',
      df.groupby('odor')['edible_flag'].mean().head(4))

# Normalizing a derived numeric frequency feature (Min-Max Scaling)
odor_counts = df['odor'].value_counts()
df['odor_frequency'] = df['odor'].map(odor_counts)

df['odor_frequency_norm'] = (
    (df['odor_frequency'] - df['odor_frequency'].min()) /
    (df['odor_frequency'].max() - df['odor_frequency'].min())
)

print('\nOriginal vs Min-Max Normalized Odor Frequency:\n',
      df[['odor_frequency', 'odor_frequency_norm']].head(3))
```
